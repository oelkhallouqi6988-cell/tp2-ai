```python
from sklearn.feature_extraction.text import CountVectorizer

# Créer une liste de 3 phrases
phrases = [
    "Le football est un sport populaire",
    "Le cinéma est un art populaire",
    "Le football et le cinéma sont intéressants"
]

# Créer le modèle Bag of Words
vectorizer = CountVectorizer()

# Transformer les phrases en matrice numérique
X = vectorizer.fit_transform(phrases)

# Afficher les mots extraits
print("Liste des mots :")
print(vectorizer.get_feature_names_out())

# Afficher la matrice
print("\nMatrice Bag of Words :")
print(X.toarray())
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# Utiliser les mêmes phrases
phrases = [
    "Le football est un sport populaire",
    "Le cinéma est un art populaire",
    "Le football et le cinéma sont intéressants"
]

# Créer le modèle TF-IDF
tfidf_vectorizer = TfidfVectorizer()

# Transformer les phrases en matrice TF-IDF
X_tfidf = tfidf_vectorizer.fit_transform(phrases)

# Afficher les mots extraits
print("Liste des mots :")
print(tfidf_vectorizer.get_feature_names_out())

# Afficher la matrice TF-IDF
print("\nMatrice TF-IDF :")
print(X_tfidf.toarray().round(3))
```
